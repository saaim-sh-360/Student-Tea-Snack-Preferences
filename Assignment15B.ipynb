```python
# Assignment 15B: Student Tea & Snack Preferences
# Importing libraries and reading the dataset
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\Administrator\Downloads\student_tea_snacks _1_.csv")
# Q1: What is the most preferred tea type?
print("Q1:\n", df.groupby('tea_type')['student_id'].count().sort_values(ascending=False).head(1))
# Q2: What is the average number of cups per gender?
print("Q2:\n", round(df.groupby('gender')['cups_per_day'].mean(), 2))
# Q3: Bar chart of snack choices.
plt.bar(df.groupby('snack_choice')['student_id'].count().index, df.groupby('snack_choice')['student_id'].count().values, color='green', width=0.5); plt.xlabel('Snack Choice'); plt.ylabel('Count'); plt.title('Snack Choices'); plt.show()
# Q4: Compare tea preference by age group.
print("Q4:\n", df.groupby(['age_group', 'prefers_tea'])['student_id'].count())
# Q5: Number of students who don't drink tea.
print("Q5:", len(df[df['prefers_tea'] == 'No']))
```
